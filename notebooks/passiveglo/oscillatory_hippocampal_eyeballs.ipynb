```python
import collections
import functools
import logging
import numpy as np
import os
import quantities as pq

import epych
from epych.statistics import alignment, spectrum
```

```python
%matplotlib inline
```

```python
logging.basicConfig(level=logging.INFO)
```

```python
ADAPTED_ONSET = pq.Quantity(-1.0) * pq.second
ADAPTED_OFFSET = pq.Quantity(-0.5) * pq.second
ODDBALL_ONSET = pq.Quantity(-1.9017372477960602e-14) * pq.second
ODDBALL_OFFSET = pq.Quantity(0.5004545430388676) * pq.second
EVENTS = {
    "Adapted": (ADAPTED_ONSET.magnitude, 'lightgreen'),
    "Adapted Offset": (ADAPTED_OFFSET.magnitude, 'red'),
    "Oddball": (ODDBALL_ONSET.magnitude, 'lightgreen'),
    "Offset": (ODDBALL_OFFSET.magnitude, 'red'),
}
FREQ_RANGE = (spectrum.THETA_BAND[0], spectrum.GAMMA_BAND[1])
```

```python
CONDITIONS = ["lonaive", "go_gloexp", "go_seqctl", "lo_gloexp", "lo_rndctl", "igo_seqctl"]
CONDITION_TITLES = {
    "lonaive": "xxxy (Habituation)",
    "go_gloexp": "xxxX",
    "go_seqctl": "xxxx",
    "lo_gloexp": "xxxY",
    "lo_rndctl": "XXXY",
    "igo_seqctl": "yyyy"
}
```

```python
area_titles = {
    "VISal": "AL",
    "VISam": "AM",
    "VISl": "LM",
    "VISp": "V1",
    "VISpm": "PM",
    "VISrl": "RL",
}
def stattitle(name, stat):
    if name in area_titles:
        return "Sub-" + area_titles[name]
    return name
anatomical_areas = ['VISp', 'VISl', 'VISrl', 'VISal', 'VISpm', 'VISam']
```

```python
def condition_stattitle(cond, name, stat):
    return stattitle(name, stat) + (" (%s)" % CONDITION_TITLES[cond])
```

```python
def initialize_spectrum(key, signal, path=None):
    area = os.path.commonprefix([loc for loc in signal.channels.location])
    return epych.statistics.spectrum.Spectrogram(signal.df, signal.channels, signal.f0, taper="hann", path=path + "/" + key, time_window=0.200)
```

```python
summaries = {}
```

```python
for cond in CONDITIONS:
    cond_path = "/mnt/data/000253/grand_hippocampal_spectrogram_downsample4_%s" % cond
    summaries[cond] = epych.statistic.Summary.unpickle(cond_path, epych.statistics.spectrum.Spectrogram).results()
    logging.info("Loaded LFP spectrograms across condition %s" % cond)
```

```python
for cond in CONDITIONS:
    cond_stattitle = functools.partial(condition_stattitle, cond)
    tfrs = summaries[cond].smap(lambda sig: sig.oscillatory().select_freqs(*FREQ_RANGE)).erp().smap(lambda tfr: tfr.decibels())
    tfrs.plot(signals=anatomical_areas, sigtitle=cond_stattitle, title="%s spectrograms by area" % cond, baseline=(-3.5, -3.0),
              figure="hippocampal_oscillatory_downsample4_%s.pdf" % cond, vmin=-7.5, vmax=7.5, **EVENTS)
    for area in anatomical_areas:
        os.makedirs("hippocampal_oscillatory_downsample4_%s/" % cond, exist_ok=True)
        tfrs.signals[area].plot(title=cond_stattitle(area, tfrs.signals[area]), filename="hippocampal_oscillatory_downsample4_%s/%s.pdf" % (cond, area),
                                baseline=(-3.5, -3.0), vmin=-7.5, vmax=7.5, cbar_ends=("Stronger", "Weaker"), tlabel="Time from Oddball (seconds)", **EVENTS)
        oddball_signal = tfrs.signals[area][ADAPTED_ONSET.magnitude - 0.25:ODDBALL_OFFSET.magnitude + 0.25]
        oddball_signal.plot(title=cond_stattitle(area, tfrs.signals[area]), vmin=-7.5, vmax=7.5,
                            filename="hippocampal_oscillatory_downsample4_%s/%s_oddball.pdf" % (cond, area), cbar_ends=("Stronger", "Weaker"), **EVENTS)
```
